## Q6. Plotting
Create a line plot of $y = x^2$ for $x$ values from 0 to 10.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0,10)
y = x**2

plt.plot(x, y, label="y=x**2")
plt.title("Plot Quiz")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
