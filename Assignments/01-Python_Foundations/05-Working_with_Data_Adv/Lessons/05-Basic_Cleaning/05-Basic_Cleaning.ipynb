# 🧼 Cleaning Dirty Data (Missing Values & Type Fixes)

## 🔹 LEARNING GOALS:
- Detect and count missing values (`NaN`)
- Fill or drop missing data
- Convert column data types safely
- Understand the difference between `NaN`, `None`, `""`, and type mismatches


### 🧪 1. Load a Messy Dataset

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", None],
    "Age": ["25", "thirty", 35, np.nan, "40"],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022"],
    "Score": [95.5, None, 88.0, 92.5, ""]
}

df = pd.DataFrame(data)
df
```

### 🧯 2. Detecting Missing or Broken Values

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df[df.isnull().any(axis=1)]
```

### 🧹 3. Cleaning Strategy Options

```python
df.fillna({
    "Name": "Unknown",
    "Age": -1,
    "Signup Date": "1970-01-01",
    "Score": 0.0
})
```

```python
df.dropna()
```

### 🧬 4. Data Type Fixes

```python
df.dtypes
```

```python
df["Age"] = pd.to_numeric(df["Age"], errors="coerce")
df["Score"] = pd.to_numeric(df["Score"], errors="coerce")
df["Signup Date"] = pd.to_datetime(df["Signup Date"], errors="coerce")
df.dtypes
```

### 🩹 5. Impute (Fill In) Fixed Missing Values

```python
df["Age"].fillna(df["Age"].median(), inplace=True)
df["Score"].fillna(df["Score"].mean(), inplace=True)
```

```python
df["Signup Date"].fillna(df["Signup Date"].min(), inplace=True)
```

```python
df["Name"].fillna("Unknown", inplace=True)
```

### 🤓 6. Cleaned Data Review

```python
df.info()
```

```python
df.describe(include="all")
```

```python
myData = {
    "Name": ["Alice", "Bob", "Charlie", "David", None, None],
    "Age": ["25", "thirty", 35, np.nan, "40", None],
    "Signup Date": ["2022-01-01", "not a date", "2022/03/01", None, "April 5, 2022", None],
    "Score": [95.5, None, 88.0, 92.5, "", None],
    "Height": [None, None, None, None, None, None]
}

md = pd.DataFrame(myData)
md
md.isnull()
md.isnull().sum()
md[md.isnull().any(axis=1)]

md.fillna({
    "Name": "Unknown",
    "Age": -1,
    "Signup Date": "1970-01-01",
    "Score": 0.0,
    "Height": -1
})

md.dropna()

md.dtypes

md["Age"] = pd.to_numeric(md["Age"], errors="coerce")
md["Score"] = pd.to_numeric(md["Score"], errors="coerce")
md["Signup Date"] = pd.to_datetime(md["Signup Date"], errors="coerce")
md["Height"] = pd.to_numeric(md["Height"], errors="coerce")
md.dtypes

md["Age"].fillna(md["Age"].median(), inplace=True)
md["Score"].fillna(md["Score"].mean(), inplace=True)

md["Signup Date"].fillna(md["Signup Date"].min(), inplace=True)

md["Name"].fillna("Unknown", inplace=True)
md["Height"].fillna(md["Height"].median(), inplace=True)

md.info()

df.describe(include="all")
```
